```python
import pandas as pd
```

```python
ecommerce = pd.read_csv(r"C:\Users\Asus vivobook\Downloads\pakistani_ecommerce_dataset.csv")
```

```python
ecommerce.head()
```

```python
ecommerce.columns
```

```python
ecommerce.info()
```

```python
category_counts = ecommerce["product_category"].value_counts()
category_counts
```

```python
most_popular_category = category_counts.idxmax()
most_popular_category
```

```python
average_order_value = ecommerce["order_value"].mean()
round(average_order_value, 2)
```

```python
city_orders = ecommerce["city"].value_counts()
city_orders
```

```python
top_city = city_orders.idxmax()
top_city
```

```python
payment_counts = ecommerce["payment_method"].value_counts()
payment_counts
```

```python
most_used_payment = payment_counts.idxmax()
most_used_payment
```

```python
average_by_city = ecommerce.groupby("city")["order_value"].mean()
average_by_city
```

```python
average_by_city.plot(
    kind="bar",
    color="green",
    width=0.5,
    xlabel="Cities",
    ylabel="Average Order Value",
    title="Average Order Value by City",
    rot=45
)
```

```python
students = pd.read_csv(r"C:\Users\Asus vivobook\Downloads\student_lifestyle_dataset.csv")
```

```python
students.head()
```

```python
students.columns
```

```python
students.info()
```

```python
average_study_hours = students["study_hours"].mean()
round(average_study_hours, 2)
```

```python
average_sleep = students.groupby("gender")["sleep_hours"].mean()
average_sleep
```

```python
gender_with_more_sleep = average_sleep.idxmax()
gender_with_more_sleep
```

```python
screen_sleep_correlation = students["screen_time"].corr(students["sleep_hours"])
round(screen_sleep_correlation, 3)
```

```python
students.plot(
    kind="scatter",
    x="screen_time",
    y="sleep_hours",
    title="Screen Time and Sleep Hours",
    xlabel="Screen Time",
    ylabel="Sleep Hours"
)
```

```python
students["daily_transport_cost"].plot(
    kind="hist",
    bins=20,
    color="blue",
    edgecolor="black",
    alpha=0.7,
    title="Histogram of Daily Transport Cost",
    xlabel="Daily Transport Cost",
    ylabel="Frequency"
)
```

```python
average_internet = students.groupby("gender")["internet_usage"].mean()
average_internet
```

```python
average_internet.plot(
    kind="bar",
    color="green",
    width=0.5,
    xlabel="Gender",
    ylabel="Average Internet Usage",
    title="Average Internet Usage by Gender",
    rot=0
)
```
